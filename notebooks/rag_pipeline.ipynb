# 2.1 Load & Inspect

In this section, I load the collected PDF documents, inspect
their page counts, extract their text, and verify that the
documents are text-extractable and do not require OCR.

```python
from pathlib import Path
from pypdf import PdfReader
import re

DATA_FOLDER = Path("../Data")

pdf_files = sorted(DATA_FOLDER.glob("*.pdf"))

print("Number of PDF files:", len(pdf_files))
print()

documents = []

# Outer Loop: Iterates over each document file
for pdf in pdf_files:

    reader = PdfReader(pdf)

    print("=" * 60)
    print("FILE:", pdf.name)
    print("NUMBER OF PAGES:", len(reader.pages))

    # Inner Loop: Iterates over each individual page inside that document
    for page_number, page in enumerate(reader.pages, start=1):

        text = page.extract_text() or ""

        # Basic cleaning
        text = re.sub(r"\s+", " ", text).strip()

        if text:
            documents.append({
                "text": text,
                "source": pdf.name,
                "page": page_number
            })

print()
print("=" * 60)
print("TOTAL EXTRACTED PAGES:", len(documents))
```

```python
print("\nSample extracted pages:\n")

for document in documents[:3]:

    print("-" * 60)
    print("Source:", document["source"])
    print("Page:", document["page"])
    print("Text:")
    print(document["text"][:500])
    print()
```

# 2.2 Chunking Strategy
I am using fixed-size chunking with overlap.
Chunk size: 1000 words
Overlap: 200 words
Each document is divided into chunks of approximately 1000 words,
with an overlap of 200 words between consecutive chunks.
The overlap helps preserve context when important information
appears near the boundary between two chunks.

```python
def fixed_size_chunk(page_text, chunk_size=1000, overlap=200):
    words = page_text.split()
    chunks = []
    step=chunk_size - overlap            #every 800 word start new chunk
    
    for i in range(0, len(words), step):
        chunk = " ".join(words[i:i + chunk_size])   
        if chunk:
            chunks.append(chunk)              
        
    return chunks
```

```python
all_chunks=[]
for document in documents:
    page_text=document["text"]      #text belonging to this one page
    chunks=fixed_size_chunk(page_text,chunk_size=1000, overlap=200)       # chunks=[chunk 1,chunk2,chunk3,chunk4,...]
    for chunk in chunks:
        all_chunks.append({                                              #all_chunks-->record
            "text":chunk,
            "source":document["source"],
            "page":document["page"]
        })
print("Total chunks created:", len(all_chunks))
for i in range(min(3, len(all_chunks))):

    print("=" * 60)
    print("CHUNK:", i + 1)
    print("SOURCE:", all_chunks[i]["source"])
    print("PAGE:", all_chunks[i]["page"])
    print("TEXT:")
    print(all_chunks[i]["text"][:500])
    print()
```

# 2.3  Embeddings & Vector Store
i used the `all-MiniLM-L6-v2` embedding model

```python
from sentence_transformers import SentenceTransformer
import chromadb

embedding_model = SentenceTransformer("all-MiniLM-L6-v2")
print("Embedding model loaded successfully.")
```

```python
chroma_client = chromadb.PersistentClient(   #Save the database to disk
    path="../vector_store")
collection = chroma_client.get_or_create_collection(
    name="cs_learning_documents") #cs_learning_documents is like database or table contain chunks
print("Chroma collection created successfully.")
```

```python
chunk_texts=[]
ids=[]
metadatas=[]
for i, chunk in enumerate(all_chunks):
    chunk_texts.append(chunk["text"])     
    ids.append(f"chunk_{i}")
    metadatas.append({
        "source": all_chunks[i]["source"],
        "page": all_chunks[i]["page"]
    })
print("Number of texts:", len(chunk_texts))
```

```python
embeddings = embedding_model.encode(chunk_texts)

print("Embeddings generated successfully.")
print("Number of embeddings:", len(embeddings))
print("Embedding size:", len(embeddings[0]))
```

```python
collection.add(
    ids=ids,
    documents=chunk_texts,
    embeddings=embeddings.tolist(),
    metadatas=metadatas
)

print("Chunks added to Chroma successfully.")
print("Total documents in collection:", collection.count())
```

# 2.4 Retrieval & Prompting

In this section, i retrieve the most relevant chunks from the
Chroma vector store based on a user's question.

The user's question is converted into an embedding using the same
embedding model used for the document chunks.

Chroma then compares the question embedding with the stored
embeddings and returns the most relevant chunks.

These retrieved chunks will later be provided to the language model
as context so that the generated answer is grounded in the
collected documents.

```python
def find_relevant_chunks(question, n_results=3):
    question_embedding=embedding_model.encode([question])     #we made[question] bec. it means a list containing one question bec. embedding model expects a list
    results=collection.query(query_embeddings=question_embedding.tolist(), n_results=n_results)
    return results
```

```python
test_questions = [
    "What is machine learning?",
    "What is the difference between artificial intelligence and machine learning?",
    "What is overfitting?",
    "What is stochastic gradient descent?",
    "What is a neural network?",
    "What is artificial intelligence?",
    "What is the Turing test?",
    "What is the distributional hypothesis?",
    "What are word embeddings?",
    "What is vector semantics?"
]
```

```python
for question in test_questions:
    results = find_relevant_chunks(question, n_results=3)

    print("=" * 60)
    print("QUESTION:", question)

    for i in range(len(results["documents"][0])):

        print("RESULT:", i + 1)
        print("SOURCE:", results["metadatas"][0][i]["source"])
        print("PAGE:", results["metadatas"][0][i]["page"])
```

## Prompt Template

The retrieved chunks are combined into a context that is provided
to the language model together with the user's question.

The prompt instructs the language model to answer using only the
retrieved context. If the answer is not available in the provided
context, the model should say that it could not find the answer
in the provided documents.

The prompt also asks the model to provide the source document and
page number used for the answer.

```python
def build_prompt(question, results):
    context = ""
    for i in range(len(results["documents"][0])):
        source = results["metadatas"][0][i]["source"]
        page = results["metadatas"][0][i]["page"]
        text = results["documents"][0][i]

        context += f"\nSource: {source}, Page: {page}\n"
        context += f"{text}\n"

    prompt = f"""
You are an AI and Computer Science Learning Assistant.
Answer the user's question using ONLY the provided context.
If the answer cannot be found in the context, say:
"I couldn't find the answer in the provided documents."
Answer the question clearly and concisely.
Do not invent information or citations.
Context:
{context}

Question:
{question}

Answer:
"""

    return prompt
```

```python
question = "What is machine learning?"
results = find_relevant_chunks(question, n_results=3)
prompt = build_prompt(question, results)
print(prompt)
```

```python
def get_sources(results):
    sources = []
    for i in range(len(results["documents"][0])):
        source = results["metadatas"][0][i]["source"]
        page = results["metadatas"][0][i]["page"]
        citation = f"{source}, page {page}"
        if citation not in sources:
            sources.append(citation)

    return sources
```

```python
import ollama
def generate_answer(question, n_results=3):
    results = find_relevant_chunks(question, n_results)
    prompt = build_prompt(question, results)
    response = ollama.chat(
        model="llama3.2:3b",
        messages=[
            { "role": "user","content": prompt}])
    answer = response["message"]["content"]
    sources = get_sources(results)
    return answer, sources
```

```python
#test
question = "What is machine learning?"
answer, sources = generate_answer(question)
print("ANSWER:")
print(answer)
print("\nSOURCES:")
for source in sources:
    print("-", source)
```

# 2.6 Evaluation

I evaluate the RAG system using the same 10 test questions used
during retrieval testing.

For each question, I check whether the retrieved context is relevant
and whether the generated answer is supported by the retrieved
documents.

I also record the retrieved source, generated answer, and whether
the answer is correct.

```python
evaluation_results = []
for question in test_questions:
    results = find_relevant_chunks(question, n_results=3)
    answer, sources = generate_answer(question, n_results=3)
    evaluation_results.append({
        "question": question,
        "retrieved_source": "; ".join(sources),
        "answer": answer,
        "relevant": "",
        "grounded": "",
        "correct": ""
    })

print("Evaluation completed.")
```

## Failure Cases and Mitigation

During evaluation, some retrieved chunks may be less relevant than
others because semantic similarity does not always guarantee that
every retrieved chunk directly answers the question.

To mitigate this, the system retrieves multiple chunks and provides
them together as context to the language model. The prompt also
instructs the model to use only the retrieved context and to state
when the answer cannot be found in the provided documents.
